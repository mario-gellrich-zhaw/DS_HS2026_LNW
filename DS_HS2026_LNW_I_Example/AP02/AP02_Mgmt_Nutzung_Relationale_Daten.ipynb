## <b>Vorbereitung (Dieser Teil wird <u>nicht</u> bewertet!)</b>

#### <b>1.) Erstellen Sie eine Verbindung zur SQLite-Datenbank 'cinema.db'.</b>

```python
# Engine für SQLite-Datenbankverbindung erstellen
engine = create_engine('sqlite:///cinema.db')

# Test der Datenbankverbindung
with engine.connect() as connection:
    result = connection.execute(text('SELECT sqlite_version()'))
    print('SQLite Version:', result.fetchone()[0])

engine.dispose()
```

#### <b>2.) Lesen Sie die Tabellenblätter aus 'cinema_data.xlsx' ein und speichern Sie diese in separaten DataFrames.</b>

```python
# Pfad zur .xlsx Datei
path = 'cinema_data.xlsx'

# Einlesen der Daten
df_movies     = pd.read_excel(path, sheet_name='movies')
df_cinemas    = pd.read_excel(path, sheet_name='cinemas')
df_staff      = pd.read_excel(path, sheet_name='staff')
df_screenings = pd.read_excel(path, sheet_name='screenings')
df_customers  = pd.read_excel(path, sheet_name='customers')
df_tickets    = pd.read_excel(path, sheet_name='tickets')

# Spaltennamen zu lowercase
for df in [df_movies, df_cinemas, df_staff, df_screenings, df_customers, df_tickets]:
    df.columns = df.columns.str.lower()

print('Tabellen erfolgreich eingelesen.')
```

#### <b>3.) Verschaffen Sie sich einen Überblick über die Datenbank: Tabellen, Spalten und Zeilenanzahl.</b>

<b>Hinweis:</b> Das ER-Diagramm (cinema_er_diagram.png) gibt Ihnen eine Übersicht über das Datenbankschema.

```python
# Überblick über alle Tabellen
engine = create_engine('sqlite:///cinema.db')

tables = ['movies', 'cinemas', 'staff', 'screenings', 'customers', 'tickets']
for tbl in tables:
    df = pd.read_sql_query(f'SELECT * FROM {tbl} LIMIT 3', con=engine)
    print(f'\n--- {tbl} ({pd.read_sql_query(f"SELECT COUNT(*) AS n FROM {tbl}", con=engine)["n"][0]} Zeilen) ---')
    print(df.to_string(index=False))

engine.dispose()
```

#### <b>Hinweise zu SQLite</b>

<ul>
  <li>Jahr aus einem Datum: <code>CAST(strftime('%Y', screening_datetime) AS INTEGER)</code> statt <code>EXTRACT(YEAR FROM ...)</code></li>
  <li>Jahr-Monat: <code>strftime('%Y-%m', screening_datetime)</code> liefert z.B. <code>'2024-03'</code></li>
  <li>Datumsbereiche: <code>DATE(screening_datetime) BETWEEN '2024-01-01' AND '2024-12-31'</code></li>
  <li>Runden: <code>ROUND(AVG(paid_price), 2)</code> funktioniert direkt.</li>
