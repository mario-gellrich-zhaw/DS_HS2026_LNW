## Setup (wird nicht bewertet)

````python
# Libraries
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
from sklearn import tree
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor, DecisionTreeClassifier
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import r2_score, accuracy_score, confusion_matrix, RocCurveDisplay

# Ignore warnings
import warnings
warnings.filterwarnings('ignore')

# Get current working directory
print(os.getcwd())
````

### Velodaten laden → `df`

````python
# Import data from csv to pandas dataframe named 'df'
df = pd.read_csv('bicycle_data.csv')

# Show dimensions and first rows
print('Number of rows and columns:', df.shape)
df.head()
````

### Trainings- und Testdaten erstellen

Der folgende Code teilt die Daten **zweimal** in 80 % Trainingsdaten und 20 % Testdaten auf:

- **Preis (Regression):** `X_train`, `X_test`, `y_train`, `y_test` mit den Features in `price_features` und der Zielvariable `price_chf`.
- **Service (Klassifikation):** `X_train_class`, `X_test_class`, `y_train_class`, `y_test_class` mit den Features in `service_features` und der Zielvariable `service_needed`. `stratify` sorgt dafür, dass der Anteil der Velos mit Service in Trainings- und Testdaten gleich gross ist.

````python
# Features and target for the price prediction (regression)
price_features = ['age_years', 'mileage_km', 'equipment_level', 'is_ebike']
X_train, X_test, y_train, y_test = train_test_split(df[price_features],
                                                    df['price_chf'],
                                                    test_size=0.20,
                                                    random_state=42)

# Features and target for the service prediction (classification)
service_features = ['brake_wear_pct', 'chain_wear_pct', 'months_since_service']
X_train_class, X_test_class, y_train_class, y_test_class = train_test_split(df[service_features],
                                                                            df['service_needed'],
                                                                            test_size=0.20,
                                                                            random_state=42,
                                                                            stratify=df['service_needed'])

# Show the size of the samples
print('Regression:     ', X_train.shape[0], 'Trainingsdaten /', X_test.shape[0], 'Testdaten')
print('Klassifikation: ', X_train_class.shape[0], 'Trainingsdaten /', X_test_class.shape[0], 'Testdaten')
````
