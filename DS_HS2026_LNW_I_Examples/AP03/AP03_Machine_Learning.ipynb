| `months_since_service` | Monate seit letztem Service | Servicemerkmal |
| `price_chf` | Später erzielter Verkaufspreis in CHF | Regressionsziel |
| `service_needed` | Späterer Werkstattentscheid: 1 Service nötig, 0 kein Service nötig | Klassifikationsziel |

````python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.tree import DecisionTreeRegressor, DecisionTreeClassifier
from sklearn.ensemble import (
    RandomForestRegressor, GradientBoostingRegressor,
    RandomForestClassifier, GradientBoostingClassifier,
)
from sklearn.metrics import (
    mean_squared_error, r2_score, accuracy_score, precision_score,
    recall_score, f1_score, roc_auc_score, confusion_matrix,
    ConfusionMatrixDisplay, RocCurveDisplay,
)

# Funktioniert sowohl mit dem Notebook-Ordner als auch dem Repo als Arbeitsordner.
data_candidates = [
    Path('bicycle_data.csv'),
    Path('AP03/bicycle_data.csv'),
    Path('DS_HS2026_LNW_I_Examples/AP03/bicycle_data.csv'),
]
data_path = next((p for p in data_candidates if p.is_file()), None)
if data_path is None:
    raise FileNotFoundError('bicycle_data.csv fehlt. Öffnen Sie das Notebook im AP03-Ordner.')
df = pd.read_csv(data_path)
print('Arbeitsverzeichnis:', Path.cwd())
print('Zeilen und Spalten:', df.shape)
df.head()
````

````python
price_features = ['age_years', 'mileage_km', 'equipment_level', 'is_ebike']
service_features = ['brake_wear_pct', 'chain_wear_pct', 'months_since_service']

train_reg, test_reg = train_test_split(df, test_size=0.2, random_state=42)
train_clf, test_clf = train_test_split(
    df, test_size=0.2, stratify=df['service_needed'], random_state=42)

print('Regression:     ', len(train_reg), 'Training /', len(test_reg), 'Test')
print('Klassifikation: ', len(train_clf), 'Training /', len(test_clf), 'Test')
print('Service-Anteil Training:', round(train_clf['service_needed'].mean(), 3))
print('Service-Anteil Test:    ', round(test_clf['service_needed'].mean(), 3))
````
