## Setup (wird nicht bewertet)

Die erste Code-Zelle importiert die Bibliotheken, sucht `bicycle_data.csv` und liest die Daten mit `pd.read_csv(data_path)` als DataFrame `df` ein. `df.shape` zeigt die Anzahl der Zeilen und Spalten; `df.head()` zeigt die ersten fünf Zeilen.

Die zweite Code-Zelle legt fest, welche Features für welche Aufgabe verwendet werden, und erstellt die Train / Test Splits:

| Code / Variable | Bedeutung |
|---|---|
| `price_features` | Liste der vier Spaltennamen für die Preisvorhersage: Alter, Kilometerstand, Ausstattung und E-Bike-Status. Das zugehörige Target (Regression) ist `price_chf`. |
| `service_features` | Liste der drei Spaltennamen für die Servicevorhersage: Bremsverschleiss, Kettenverschleiss und Monate seit dem letzten Service. Das zugehörige Target (Classification) ist `service_needed`. |
| `train_reg`, `test_reg` | Training Data und Test Data für Regression, erzeugt durch den ersten `train_test_split(...)`. |
| `train_clf`, `test_clf` | Training Data und Test Data für Classification, erzeugt durch den zweiten `train_test_split(...)`. |
| `test_size=0.2` | 20 % der Zeilen werden als Test Data zurückgehalten; 80 % werden als Training Data verwendet. |
| `random_state=42` | Fester Seed, damit der jeweilige Split bei erneutem Ausführen mit denselben Daten reproduzierbar bleibt. |
| `stratify=df['service_needed']` | Der Classification Split erhält die Anteile der beiden Classes `0` und `1` in Training Data und Test Data annähernd gleich. |

`price_features` und `service_features` enthalten **Spaltennamen**, noch keine Daten. Zum Beispiel wählt `train_reg[price_features]` die Feature-Spalten aus den Training Data aus und bildet damit `X`; `train_reg['price_chf']` wählt das Target und bildet `y`. Für Classification gilt entsprechend `train_clf[service_features]` für `X` und `train_clf['service_needed']` für `y`.

Die Splits teilen **Zeilen** auf; alle Spalten bleiben zunächst enthalten. Deshalb werden Features und Target erst in der jeweiligen Aufgabe getrennt ausgewählt. Die Targets gehören nicht in `X`. Die beiden Splits können wegen `stratify` unterschiedliche Zeilen enthalten.

Die letzten `print(...)`-Zeilen zeigen die Anzahl der Zeilen je Split und den Anteil von `service_needed = 1`. Da dieses Target nur `0` und `1` enthält, entspricht `.mean()` dem Anteil der positiven Class («Service nötig»).


````python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.tree import DecisionTreeRegressor, DecisionTreeClassifier
from sklearn.ensemble import (
    RandomForestRegressor, GradientBoostingRegressor,
    RandomForestClassifier, GradientBoostingClassifier,
)
from sklearn.metrics import (
    mean_squared_error, r2_score, accuracy_score, precision_score,
    recall_score, f1_score, roc_auc_score, confusion_matrix,
    ConfusionMatrixDisplay, RocCurveDisplay,
)

# Funktioniert sowohl mit dem Notebook-Ordner als auch dem Repo als Arbeitsordner.
data_candidates = [
    Path('bicycle_data.csv'),
    Path('AP03/bicycle_data.csv'),
    Path('DS_HS2026_LNW_I_Examples/AP03/bicycle_data.csv'),
]
data_path = next((p for p in data_candidates if p.is_file()), None)
if data_path is None:
    raise FileNotFoundError('bicycle_data.csv fehlt. Öffnen Sie das Notebook im AP03-Ordner.')
df = pd.read_csv(data_path)
print('Arbeitsverzeichnis:', Path.cwd())
print('Zeilen und Spalten:', df.shape)
df.head()
````

````python
# Feature-Spaltennamen fuer Regression; Target: price_chf
price_features = ['age_years', 'mileage_km', 'equipment_level', 'is_ebike']

# Feature-Spaltennamen fuer Classification; Target: service_needed
service_features = ['brake_wear_pct', 'chain_wear_pct', 'months_since_service']

# Regression: Zeilen in 80 % Training Data und 20 % Test Data aufteilen.
train_reg, test_reg = train_test_split(df, test_size=0.2, random_state=42)

# Classification: gleicher Anteil der Classes dank stratify (stratified split).
train_clf, test_clf = train_test_split(
    df, test_size=0.2, stratify=df['service_needed'], random_state=42)

# Split-Groessen und Anteil der positiven Class zur Kontrolle ausgeben.
print('Regression:     ', len(train_reg), 'Training /', len(test_reg), 'Test')
print('Classification: ', len(train_clf), 'Training /', len(test_clf), 'Test')
print('Service-Anteil Training:', round(train_clf['service_needed'].mean(), 3))
print('Service-Anteil Test:    ', round(test_clf['service_needed'].mean(), 3))
````
