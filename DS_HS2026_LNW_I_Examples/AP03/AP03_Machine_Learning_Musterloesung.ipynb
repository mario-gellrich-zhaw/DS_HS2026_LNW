### Trainings- und Testdaten erstellen

Der folgende Code teilt die Daten **zweimal** in 80 % Trainingsdaten und 20 % Testdaten auf:

Wie viel Prozent der Variabilität in den Trainingsdaten erklärt das Modell?

```python
# Create training and test data
X_train_B1, X_test_B1, y_train_B1, y_test_B1 = train_test_split(df[['age_years', 'mileage_km', 'equipment_level', 'is_ebike', 'price_chf']],
```

**Begründung der Änderungen:**


1. **Zielvariable aus den Features entfernen:** `price_chf` ist die Zielvariable und darf nicht gleichzeitig Feature sein. Das Modell kopiert sonst einfach den Preis (Koeffizient 1.0, alle anderen 0) — R² = 1 ist bedeutungslos. Zudem ist der Preis bei Eingang des Velos noch gar nicht bekannt.
2. **Konstante hinzufügen:** `sm.OLS()` schätzt **keinen** Achsenabschnitt, wenn er nicht mit `sm.add_constant()` ergänzt wird (in der Ausgabe fehlt `const`). Die Regressionsgerade wird dann durch den Nullpunkt gezwungen.

Ergebnis: Das korrigierte Modell erklärt rund **96.6 %** der Variabilität der Preise in den Trainingsdaten (R² = 0.966).

**Korrigierter Code:**

````python
# Create training and test data (without the target variable 'price_chf' as feature)
X_train_B1, X_test_B1, y_train_B1, y_test_B1 = train_test_split(df[['age_years', 'mileage_km', 'equipment_level', 'is_ebike']],
                                                                df['price_chf'],
                                                                test_size=0.20,
                                                                random_state=42)

# Fit the multiple linear regression model (note the 'add_constant' function)
olsres_B1 = sm.OLS(y_train_B1, sm.add_constant(X_train_B1)).fit()

# Show R-squared and coefficients
print('R-squared:', round(olsres_B1.rsquared, 3))
print(olsres_B1.params.round(2))
print(f'\nDas Modell erklärt rund {olsres_B1.rsquared * 100:.1f} % der Variabilität in y_train_B1.')
````

```python
# Create training and test data

                                                                df['service_needed'],
                                                                test_size=0.80,
                                                                random_state=42,

# Create and train the random forest classifier
```

Tatsächliches Ergebnis:

```
Accuracy: 1.0

 [  0  59]]
```

Der Datensatz enthält **1'200 Velos**. Der Analyst meldet: *«Das Modell erkennt jeden Servicefall fehlerfrei.»*

**Aufgabe:** Schreiben Sie den korrigierten Code in die Zelle «Korrigierter Code» (die Kommentare des ursprünglichen Codes sind als Gerüst vorgegeben). Begründen Sie jede Änderung gegenüber dem ursprünglichen Code und erklären Sie, warum das ursprüngliche Ergebnis zu optimistisch ist.

**Begründung der Änderungen:**


1. **Aufteilung 80 / 20:** `test_size=0.80` legt 80 % der Daten in die **Test**daten; trainiert wurde nur mit 240 Velos (181 + 59). Richtig ist `test_size=0.20`.
2. **Auswertung auf den Testdaten:** Vorhersage und Kennzahlen wurden mit den **Trainings**daten berechnet. Diese Velos kennt das Modell bereits — der Random Forest hat sie «auswendig gelernt» (Overfitting). Für neue Velos sagt das nichts aus. Richtig: `predict(X_test_B2)` und Vergleich mit `y_test_B2`.

Ergebnis: Accuracy **0.887** auf 240 Testvelos. Von 59 Velos mit Servicebedarf werden 43 erkannt und 16 übersehen (Recall = 43 / 59 = 0.73) — gut, aber nicht fehlerfrei.

**Korrigierter Code:**

````python
# Create training and test data (20 % test data)
X_train_B2, X_test_B2, y_train_B2, y_test_B2 = train_test_split(df[['brake_wear_pct', 'chain_wear_pct', 'months_since_service']],
                                                                df['service_needed'],
                                                                test_size=0.20,
                                                                random_state=42,
                                                                stratify=df['service_needed'])

# Create and train the random forest classifier
rfc_B2 = RandomForestClassifier(n_estimators=100, random_state=42)
rfc_B2.fit(X_train_B2, y_train_B2)

# Predict the target variable for the test data
y_pred_B2 = rfc_B2.predict(X_test_B2)

# Evaluate the model on the test data
print('Accuracy:', round(accuracy_score(y_test_B2, y_pred_B2), 3))
print('Confusion matrix:')
print(confusion_matrix(y_test_B2, y_pred_B2))
````
