### Produktlisten-Seiten scrapen → `df_raw`

Der folgende Code liest die drei gespeicherten Produktlisten-Seiten mit BeautifulSoup ein. Jede Produktkarte (`<div class="product-card">`) wird zu einer Zeile im Data Frame `df_raw`. Die Werte werden **als Text** übernommen (Spalten mit Endung `_raw`) und noch nicht bereinigt.

````python
rows = []
for page in [1, 2, 3]:
    with open(f'html/shop_page_{page}.html', encoding='utf-8') as f:
        soup = BeautifulSoup(f.read(), 'html.parser')

    for card in soup.find_all('div', class_='product-card'):
        rows.append({
            'product_id':   card['data-product-id'],
            'brand':        card.find('span', class_='brand').get_text(strip=True),
            'model':        card.find('span', class_='model').get_text(strip=True),
            'storage_raw':  card.find('li', class_='storage').get_text(strip=True),
            'color':        card.find('li', class_='color').get_text(strip=True),
            'screen_raw':   card.find('li', class_='screen').get_text(strip=True),
            'battery_raw':  card.find('li', class_='battery').get_text(strip=True),
            'price_raw':    card.find('span', class_='price').get_text(strip=True),
            'availability': card.find('span', class_='availability').get_text(strip=True),
        })

df_raw = pd.DataFrame(rows)
print('Zeilen:', df_raw.shape[0], '| Spalten:', df_raw.shape[1])
df_raw.head()
````
