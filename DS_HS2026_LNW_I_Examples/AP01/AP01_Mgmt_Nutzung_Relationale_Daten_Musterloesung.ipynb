## Teil B – SQL-Fehleranalyse (16 Punkte)

*Jede Abfrage läuft fehlerfrei durch, liefert aber ein falsches oder irreführendes Geschäftsergebnis. Die tatsächlichen Ergebnisse aus `cinema.db` sind angegeben.*

### B1 (8 Punkte)

**Ziel:** Anzahl verkaufter Tickets pro Kino — für **alle** Kinos in der Datenbank.

```sql
SELECT c.name, COUNT(*) AS ticket_count
FROM screenings s
JOIN tickets t ON s.screeningid = t.screeningid
JOIN cinemas c ON s.cinemaid = c.cinemaid
GROUP BY c.name;
```

Die Tabelle `cinemas` enthält **6 Kinos**. Tatsächliches Ergebnis der SQL-Abfrage: nur **5 Zeilen**, Ticketanzahl zwischen 350 und 450.

Das Management fragt: *«Zeig mir auch Kinos, die noch keine Tickets verkauft haben.»*

**Aufgabe:** Schreiben Sie die korrigierte SQL-Abfrage, sodass **alle 6 Kinos** erscheinen und ein Kino ohne Tickets korrekt mit `0` ausgewiesen wird. Begründen Sie jede Änderung gegenüber der ursprünglichen SQL-Abfrage.

**Begründung der Änderungen:**

`INNER JOIN` gibt nur Zeilen zurück, für die in **beiden** Tabellen eine übereinstimmende Zeile existiert. Das Kino «Kino Aurora» hat noch keine Vorstellungen und damit auch keine Tickets — es findet keine Entsprechung in `screenings` bzw. `tickets` und fällt deshalb aus dem Ergebnis.

Zwei Änderungen sind nötig:

1. Beide `JOIN`s müssen zu `LEFT JOIN` werden, ausgehend von `cinemas` als linker Tabelle. So bleiben alle 6 Kinos erhalten.
2. `COUNT(*)` muss zu `COUNT(t.ticketid)` werden. `COUNT(*)` zählt Zeilen und würde die durch den `LEFT JOIN` erzeugte NULL-Zeile als `1` mitzählen — «Kino Aurora» erschiene fälschlich mit `ticket_count = 1`. `COUNT(t.ticketid)` ignoriert NULL-Werte und liefert korrekt `0`.

**Korrigierte SQL-Abfrage:**

````python
engine = create_engine(db_connection_url)

df_B1 = pd.read_sql_query("""
SELECT
    c.name,
    COUNT(t.ticketid) AS ticket_count
FROM cinemas c
LEFT JOIN screenings s ON c.cinemaid    = s.cinemaid
LEFT JOIN tickets t    ON s.screeningid = t.screeningid
GROUP BY c.name
ORDER BY ticket_count DESC
""", con=engine)

engine.dispose()
df_B1
````

### B2 (8 Punkte)

**Ziel:** Liste aller Kunden (`customerid`, `firstname`, `lastname`), die noch nie ein Ticket gekauft haben.

```sql
SELECT c.customerid, c.firstname, c.lastname
FROM customers c
JOIN tickets t ON c.customerid = t.customerid
WHERE t.ticketid IS NULL;
```

Tatsächliches Ergebnis: 0 Zeilen — obwohl es Kunden ohne Tickets gibt.

**Aufgabe:** Schreiben Sie die korrigierte SQL-Abfrage, die alle Kunden ohne Ticket ausgibt. Begründen Sie jede Änderung gegenüber der ursprünglichen SQL-Abfrage.

**Begründung der Änderungen:**

`INNER JOIN` liefert nur Zeilen, bei denen in beiden Tabellen eine Übereinstimmung existiert. Kunden ohne Ticket haben keinen Eintrag in `tickets` und werden durch den JOIN bereits ausgeschlossen — `WHERE t.ticketid IS NULL` kann daher nie zutreffen.

`JOIN` muss durch `LEFT JOIN` ersetzt werden. Damit bleiben alle Kunden im Ergebnis, auch jene ohne Ticket; für diese sind alle `tickets`-Spalten NULL, was der WHERE-Bedingung entspricht.

*Hinweis zur Bewertung: Die Aufgabe schreibt keine Sortierung vor. Das `ORDER BY` unten dient nur der stabilen Ausgabe — jede Reihenfolge ist korrekt.*

**Korrigierte SQL-Abfrage:**

````python
engine = create_engine(db_connection_url)

df_B2 = pd.read_sql_query("""
SELECT
    c.customerid,
    c.firstname,
    c.lastname
FROM customers c
LEFT JOIN tickets t ON c.customerid = t.customerid
WHERE t.ticketid IS NULL
ORDER BY c.customerid
""", con=engine)

engine.dispose()
df_B2
````

### C1 (4 Punkte)

Die gültigen Werte in `customers.membership_type` sind `Standard`, `Premium`, `Student` und `Senior`. Schreiben Sie eine SQL-Abfrage, die alle Kunden (`customerid`, `firstname`, `lastname`, `membership_type`) mit einem ungültigen Wert in `membership_type` ausgibt — d.h. alle Werte ausser diesen vier. Begründen Sie Ihre Filterbedingung.

**Begründung der Vorgehensweise:**

Alle geforderten Spalten liegen in `customers` — ein JOIN ist nicht nötig. `NOT IN ('Standard', 'Premium', 'Student', 'Senior')` behält nur Werte, die keinem der vier gültigen Werte exakt entsprechen. Damit werden abweichende Gross-/Kleinschreibung (`standard`, `STANDARD`) und Tippfehler (`Standart`) gleichermassen erfasst. `membership_type` enthält keine NULL-Werte, sodass `NOT IN` hier keine Zeilen verliert.

**SQL-Abfrage:**

````python
engine = create_engine(db_connection_url)

df_C1 = pd.read_sql_query("""
SELECT
    customerid,
    firstname,
    lastname,
    membership_type
FROM customers
WHERE membership_type NOT IN ('Standard', 'Premium', 'Student', 'Senior')
ORDER BY customerid
""", con=engine)

engine.dispose()
df_C1
````

### C2 (4 Punkte)

Wie hoch ist der durchschnittliche bezahlte Ticketpreis (`paid_price`) pro Mitgliedschaftstyp (`membership_type`)? Geben Sie `membership_type` und den Durchschnittspreis (`avg_price`) aus, gerundet auf 2 Dezimalstellen mit `ROUND(CAST(... AS NUMERIC), 2)`. Sortieren Sie absteigend nach `avg_price`. Begründen Sie, welche Tabellen Sie verbinden müssen und warum.

**Begründung der Vorgehensweise:**

`membership_type` liegt in `customers`, `paid_price` in `tickets`. Ein JOIN über `customerid` verbindet beide Tabellen. `GROUP BY c.membership_type` gruppiert die Tickets nach Mitgliedschaftstyp, `AVG(t.paid_price)` berechnet den Durchschnitt. `ROUND(CAST(... AS NUMERIC), 2)` rundet auf Rappen — das `CAST` ist nötig, damit die SQL-Abfrage auch unter PostgreSQL läuft, wo `ROUND` mit zwei Argumenten nur für `numeric` definiert ist.

Ein `INNER JOIN` ist hier korrekt: Kunden ohne Ticket haben keinen bezahlten Preis und dürfen den Durchschnitt nicht verfälschen.

**SQL-Abfrage:**

````python
engine = create_engine(db_connection_url)

df_C2 = pd.read_sql_query("""
SELECT
    c.membership_type,
    ROUND(CAST(AVG(t.paid_price) AS NUMERIC), 2) AS avg_price
FROM customers c
JOIN tickets t ON c.customerid = t.customerid
GROUP BY c.membership_type
ORDER BY avg_price DESC
""", con=engine)

engine.dispose()
df_C2
````

### C3 (4 Punkte)

Welche Kinos haben mehr als 400 Tickets verkauft? Geben Sie `name` und Anzahl Tickets (`ticket_count`) aus, absteigend sortiert nach `ticket_count`. Begründen Sie, welche Tabellen Sie verbinden müssen und warum.

**Begründung der Vorgehensweise:**

`tickets` ist nicht direkt mit `cinemas` verknüpft — der Weg führt über `screenings`. Zwei JOINs verbinden `cinemas → screenings → tickets`. `GROUP BY c.name` fasst alle Tickets pro Kino zusammen; `HAVING COUNT(t.ticketid) > 400` filtert nur Kinos mit mehr als 400 verkauften Tickets.

**SQL-Abfrage:**

````python
engine = create_engine(db_connection_url)

df_C3 = pd.read_sql_query("""
SELECT
    c.name,
    COUNT(t.ticketid) AS ticket_count
FROM cinemas c
JOIN screenings s ON c.cinemaid    = s.cinemaid
JOIN tickets t    ON s.screeningid = t.screeningid
GROUP BY c.name
HAVING COUNT(t.ticketid) > 400
ORDER BY ticket_count DESC
""", con=engine)

engine.dispose()
df_C3
````
